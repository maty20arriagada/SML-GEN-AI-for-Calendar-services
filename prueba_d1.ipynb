```python
#from google.colab import drive
#drive.mount('/content/drive')

import json
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Cargar modelo TinyLlama (Opción A de la guía)
modelo_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tokenizer = AutoTokenizer.from_pretrained(modelo_id)
model = AutoModelForCausalLM.from_pretrained(
    modelo_id,
    torch_dtype=torch.float16,
    device_map="auto"
)

# Cargar benchmark
with open("/frases.json", "r") as f:
    benchmark = json.load(f)

resultados = []

for frase in benchmark["frases"]:
    prompt = f"""Dada la siguiente frase sobre una reunion, extrae los campos en formato JSON:
- titulo
- fecha (formato YYYY-MM-DD, null si no especifica)
- hora_inicio (formato HH:MM 24h, null si no especifica)
- duracion_minutos (numero, null si no especifica)
- participantes (lista de nombres, [] si no especifica)
- ubicacion (string, null si no especifica)
- recordatorio_minutos (numero, null si no especifica)
- zona_horaria (string, America/Santiago por defecto)

Frase: "{frase['texto']}"

Responde SOLO con el JSON, sin texto adicional:"""

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(**inputs, max_new_tokens=256, temperature=0.1)
    respuesta = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # Extraer solo el JSON de la respuesta
    import re
    json_match = re.search(r'\{.*\}', respuesta, re.DOTALL)
    if json_match:
        try:
            json_salida = json.loads(json_match.group())
        except:
            json_salida = {"error": "JSON invalido"}
    else:
        json_salida = {"error": "No se detecto JSON"}

    resultados.append({
        "id": frase["id"],
        "texto": frase["texto"],
        "resultado_modelo": json_salida
    })

    print(f"Frase {frase['id']}: {'OK' if 'error' not in json_salida else 'ERROR'}")

# Guardar resultados
with open("resultados_tinyllama.json", "w") as f:
    json.dump(resultados, f, indent=2, ensure_ascii=False)

print("Resultados guardados en resultados_tinyllama.json")
```

```python
import json
import re
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Configuración del modelo
modelo_id = "Qwen/Qwen2-0.5B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(modelo_id)
model = AutoModelForCausalLM.from_pretrained(
    modelo_id,
    torch_dtype="auto",
    device_map="auto"
)

# Cargar benchmark
try:
    with open("frases.json", "r") as f:
        benchmark = json.load(f)
    fecha_ref = benchmark.get("metadata", {}).get("fecha_referencia", "2026-09-01")
    dia_ref = benchmark.get("metadata", {}).get("dia_referencia", "martes")
except FileNotFoundError:
    print("Error: El archivo 'frases.json' no fue encontrado.")
    benchmark = {"frases": []}
    fecha_ref, dia_ref = "2026-09-01", "martes"

resultados = []

for frase in benchmark.get("frases", []):
    prompt = f"""Hoy es {dia_ref} {fecha_ref}.
Extrae la información de la siguiente frase en un único objeto JSON.

Campos requeridos:
- titulo (string)
- fecha (YYYY-MM-DD, null si no hay)
- hora_inicio (HH:MM 24h, null si no hay)
- duracion_minutos (int, null si no hay)
- participantes (lista de strings)
- ubicacion (string, null si no hay)
- recordatorio_minutos (int, null si no hay)
- zona_horaria (usar 'America/Santiago' si no se especifica)

Frase: "{frase['texto']}"

Respuesta (SOLO JSON):"""

    messages = [
        {"role": "system", "content": "Eres un extractor de datos que solo responde con objetos JSON válidos sin explicaciones."},
        {"role": "user", "content": prompt}
    ]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    outputs = model.generate(
        **inputs,
        max_new_tokens=256,
        temperature=0.01, # Reducir creatividad para mayor precisión
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

    respuesta = tokenizer.decode(outputs[0][len(inputs.input_ids[0]):], skip_special_tokens=True).strip()

    # Extracción mejorada de JSON (maneja objetos y listas de un solo elemento)
    json_match = re.search(r'(\{.*\}|\[.*\])', respuesta, re.DOTALL)
    json_salida = {"error": "No se pudo parsear"}

    if json_match:
        try:
            raw_json = json.loads(json_match.group())
            # Si el modelo devuelve una lista, tomamos el primer elemento
            json_salida = raw_json[0] if isinstance(raw_json, list) and len(raw_json) > 0 else raw_json
        except json.JSONDecodeError:
            json_salida = {"error": "Error de formato", "raw": respuesta}

    resultados.append({
        "id": frase.get("id"),
        "texto": frase.get("texto"),
        "resultado_modelo": json_salida
    })

    status = "OK" if "error" not in json_salida else "ERROR"
    print(f"Frase {frase.get('id')}: {status}")

# Guardar resultados finales
output_file = "resultados_qwen.json"
with open(output_file, "w", encoding='utf-8') as f:
    json.dump(resultados, f, indent=2, ensure_ascii=False)

print(f"\nProceso completado. Archivo generado: {output_file}")
```
